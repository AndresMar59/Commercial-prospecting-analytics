```python
import re
import unicodedata
from urllib.parse import unquote

_LI_PROFILE = re.compile(r"linkedin\.com/in/([^/?#\s]+)")
_BASURA_FINAL = " -;,.:*!'\")]}>\\|"


def normalizar_url(url):
    """Devuelve 'linkedin.com/in/<slug>' o None si no es una URL de perfil."""
    if not isinstance(url, str):
        return None
    s = url.replace("\\/", "/").strip().lower()
    m = _LI_PROFILE.search(s)          # el subdominio (br., sv.) queda fuera del match
    if not m:
        return None
    slug = unquote(m.group(1))         # %C3%AD -> í
    slug = unicodedata.normalize("NFC", slug).strip(_BASURA_FINAL)
    return f"linkedin.com/in/{slug}" if slug else None
```

```python
def agregar_sin_duplicados(base: pd.DataFrame, nuevo: pd.DataFrame,
                           key: str = "linkedin_url_norm") -> pd.DataFrame:
    nuevo = nuevo.copy()
    nuevo[key] = nuevo["linkedin_url"].apply(normalizar_url)

    con_clave = nuevo[key].notna()
    ya_en_base = nuevo[key].isin(set(base[key].dropna()))
    repetida_en_nuevo = con_clave & nuevo.duplicated(subset=key, keep="first")
    descartar = ya_en_base | repetida_en_nuevo

    print(f"Nuevo: {len(nuevo)} | ya en base: {ya_en_base.sum()} | "
          f"repetidas dentro del nuevo: {repetida_en_nuevo.sum()} | "
          f"sin URL: {(~con_clave).sum()} | se agregan: {(~descartar).sum()}")

    return pd.concat([base, nuevo[~descartar]], ignore_index=True)
```
