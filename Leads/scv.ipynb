```python
from prospeccion.df_limpieza import load_dataframe, save_dataframe_json
import pandas as pd
import numpy as np
```

```python
BPOS = load_dataframe(filename="Libro1.xlsx")
```

```python
from prospeccion.apollo import ApolloClient, enrich_people
import sys, os
from dotenv import load_dotenv

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
load_dotenv()

apollo = os.getenv("APOLLO_API_KEY")
```

```python
Cient = ApolloClient(key=apollo)
df_1 = enrich_people(client=Cient, people=BPOS)
```

```python
# No se aplico al funcion cleaned. Lo que tengo que hacer ahorita es extrar el current job de los diccionarios de la lista de employer history

def extract_current_job(history):
    if not isinstance(history, list):
        return {}
    for job in history:
        if isinstance(job, dict) and job.get('current'):
            return job
    return {}
```

```python
current_jobs = df_1['employment_history'].apply(extract_current_job)

df_1['current_job_title'] = current_jobs.apply(lambda j: j.get('title'))
df_1['current_job_start_date'] = current_jobs.apply(lambda j: j.get('start_date'))
df_1['grade_level_job'] = current_jobs.apply(lambda j: j.get('grade_level'))
df_1['kind_job'] = current_jobs.apply(lambda j: j.get('kind'))
df_1['major_job'] = current_jobs.apply(lambda j: j.get('major'))
```

```python
df_1['departments_primary'] = df_1['departments'].str.get(0)
df_1['subdepartments_primary'] = df_1['subdepartments'].str.get(0)
df_1['functions_primary'] = df_1['functions'].str.get(0)  
```

```python
df_1.columns
```

```python
company_actual = df_1['organization'].apply(lambda x: x if isinstance(x, dict) else {})

df_1['company_name'] = company_actual.apply(lambda c: c.get('name'))
df_1['founded_year_company'] = company_actual.apply(lambda c: c.get('founded_year'))
df_1['industry_company'] = company_actual.apply(lambda c: c.get('industry'))
df_1['estimated_num_employees_company'] = company_actual.apply(lambda c: c.get('estimated_num_employees'))
df_1['city_company'] = company_actual.apply(lambda c: c.get('city'))
df_1['state_company'] = company_actual.apply(lambda c: c.get('state'))
df_1['country_company'] = company_actual.apply(lambda c: c.get('country'))
```

```python
df_1.columns
df_1 = df_1.drop(columns=['organization_id','employment_history','organization','revealed_for_current_team','subdepartments','departments'])
```
