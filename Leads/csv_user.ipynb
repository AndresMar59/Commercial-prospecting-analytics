```python
def normalizar_url(url):
    if pd.isna(url):
        return url
    url = url.replace('\\/', '/')   # arregla las barras escapadas
    return (
        url.strip()
        .lower()
        .replace('https://', '')
        .replace('http://', '')
        .replace('www.', '')
        .rstrip('/')
    )
```

```python
df1['linkedin_url_norm'] = df1['linkedin_url'].apply(normalizar_url)
df_1_preApollo['linkedin_url_norm'] = df_1_preApollo['linkedin_url'].apply(normalizar_url)
```
